# Contract Wins Dataset 2019–2026: Orders & Tenders: starter notebook

This file is a **sample**. The full data, the withheld fields and monthly updates come from the [Radar API](https://github.com/fliarbi/radar-api-mcp). This notebook loads the sample, explores it, then gets one record's full detail from the API.

```python
import glob, pandas as pd, matplotlib.pyplot as plt
path = (glob.glob('/kaggle/input/**/contract_wins_2019_2026_sample.csv', recursive=True) or ['contract_wins_2019_2026_sample.csv'])[0]
df = pd.read_csv(path, low_memory=False)
print(df.shape); df.head()
```

```python
df.describe(include="all").T.head(30)
```

```python
df['date'] = pd.to_datetime(df.date)
t = df.groupby(['year', 'client_type']).size().unstack(fill_value=0)
t.plot.bar(stacked=True, figsize=(10, 4), title='Contract wins by year and client type (sample)')
t
```

```python
df.dropna(subset=['value_usd']).sort_values('value_usd', ascending=False)[['date', 'contractor_name', 'client_name', 'contract_type', 'value_usd']].head(20)
```

```python
df.groupby('contract_type').value_usd.agg(['count', 'median', 'sum']).sort_values('count', ascending=False)
```

## Get the full record from the Radar API

The sample holds one row per contract. The API returns the full event timeline for a company, with scope, duration and every report. [Create a key](https://fliarbi.com/radar-api/developer), then add it to Kaggle **Add-ons → Secrets** as `RADAR_API_KEY`.

```python
import os, json, requests
key = os.environ.get('RADAR_API_KEY')
try:
    from kaggle_secrets import UserSecretsClient
    key = key or UserSecretsClient().get_secret('RADAR_API_KEY')
except Exception:
    pass
value = df['contractor_domain'].dropna().astype(str).iloc[0]
if key:
    r = requests.post('https://radar-api.fliarbi.com/v1/companies/events', headers={'Authorization': f'Bearer {key}'},
                      json={'domain': value}, timeout=60)
    print(json.dumps(r.json(), indent=1)[:3000])
else:
    print('Set RADAR_API_KEY to fetch the full record for', value)
```

Each row also has `fliar_profile_url`, the company's full profile. Repository, docs and MCP setup: https://github.com/fliarbi/radar-api-mcp
